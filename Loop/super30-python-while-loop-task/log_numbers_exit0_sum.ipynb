### Create a program that repeatedly asks for numbers and stops only when the user enters 0.

```python
output=[] # create an empty list called output
i=2 # initialize the variable i to 2
while i>0: #start a while loop that will run as long as i is greater than 0
    i=int(input("enter any value and log it. enter 0 for exit")) # take input from the user and convert it to an integer, then assign it to the variable i
    if i==0: # check if the value of i is equal to 0
        break # if the condition is true, exit the loop
    output.append(i) # create a string that contains the value of i and the string "output"
print(output) # print the value of output
```

### Display the sum of all previously entered numbers.

```python
print("previous output is",output) # print the value of output
total_sum=0 # initialize a variable total_sum to 0
i=0 # initialize a variable i to 0
while i < len(output): # start a while loop that will run as long as i is less than the length of the output list
    total_sum=total_sum+output[i] # add the value of output[i] to total_sum
    i =i+1 # increment the value of i by 1
print(total_sum) # print the value of total_sum
```
