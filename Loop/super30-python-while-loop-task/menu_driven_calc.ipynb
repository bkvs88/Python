Build a menu-driven calculator

1 Add

2 Subtract

3 Multiply

4 Divide

5 Exit

```python
print("Welcome to the Menu-Driven Calculator!")

while True: # keep the program running until the user chooses to exit

    print("Select an operation:") # 
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")    
    print("5. Exit")

    choice = input("Enter your choice (1-5): ") # ask the user to input their choice of operation
    print(f"You selected option {choice}.")
    if choice == '5': # if the user chooses to exit, print a goodbye message and break the loop
        print("Exiting the calculator. Goodbye!")
        break

    num1 = float(input("Enter first number: ")) # ask the user to input the first number
    num2 = float(input("Enter second number: ")) # ask the user to input the second number

    
    print(f"First number: {num1}, Second number: {num2}")

    if choice == '1': # if the user chooses addition, calculate the sum of the two numbers and print the result
        result = num1 + num2 # calculate the sum of the two numbers
        print(f"The result of addition is: {result}")
    elif choice == '2': # if the user chooses subtraction, calculate the difference of the two numbers and print the result
        result = num1 - num2 # calculate the difference of the two numbers
        print(f"The result of subtraction is: {result}")
    elif choice == '3': # if the user chooses multiplication, calculate the product of the two numbers and print the result
        result = num1 * num2 # calculate the product of the two numbers
        print(f"The result of multiplication is: {result}")
    elif choice == '4': #if the user chooses division, check if the second number is not zero, then calculate the quotient of the two numbers and print the result
        if num2 != 0: # check if the second number is not zero to avoid division by zero error
            result = num1 / num2 #  calculate the quotient of the two numbers
            print(f"The result of division is: {result}")
        else: # if the second number is zero, print an error message
            print("Error: Division by zero is not allowed.")
    else: # if the user inputs an invalid choice, print an error message
        print("Invalid input. Please select a valid operation (1-5).")
```
