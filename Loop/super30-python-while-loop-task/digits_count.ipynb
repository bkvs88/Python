### Count the number of digits in an integer.

```python
i = 0 #initialize the variable i to 0
input_number=int(input("Enter number: ")) #enter a number from the user and store it in the variable input_number
while i<len(str(input_number)): #start a while loop that will run as long as i is less than the length of the string representation of input_number
    i=i+1 #increment i by 1
print(i) # print the value of i, which is the number of digits in input_number
```
