### Calculate the sum of digits of a number.

```python
sum_of_digits=0 #initialize the variable sum_of_digits to 0
digit=0 #initialize the variable digit to 0
input_number=int(input("Enter any number to add the sum of digits")) #enter a number from the user and store it in the variable input_number
display_number=input_number #store the value of input_number in the variable display_number to display it later
while input_number>0: #start a while loop that will run as long as input_number is greater than 0
    digit=input_number%10 #get the last digit of input_number by taking the modulus of 10 and store it in the variable digit
    sum_of_digits+=digit #add the value of digit to the variable sum_of_digits
    input_number=input_number//10 #remove the last digit from input_number by performing integer division by 10
print(f" Sum of Digits: {sum_of_digits} for given number {display_number}")
```
