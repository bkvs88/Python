```python
secret_number = 7 #initialize the variable to a number
while True: # keep asking for the number until the correct one is entered
    guess = int(input("Guess the secret number between 1 to 10: ")) #ask the user to input a number
    if guess == secret_number: #check if the guessed number is equal to the secret number
        print(f"Yay!!! You guessed the correct number {guess}.") #if yes, print a success message
        break #exit the loop
    else: #if not, print a failure message
        print(f"Sorry, {guess} is not the correct number. Try again.")  # print the failure message
```
