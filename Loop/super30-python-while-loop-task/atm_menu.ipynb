### Create a basic ATM simulation where a user can repeatedly

```python
balance=10000.00 #initialize the variable balance to 10000.00
is_running=True #initialize the variable is_running to True

print("Welcome to Worlds Best Bank") # print a welcome message

while is_running: # start a while loop that will run as long as is_running is True
    print("\n =========\n ATM Menu \n =========") # print the ATM menu
    print("1. Check Balance") # print the option to check balance
    print("2. Deposit Money") # print the option to deposit money
    print("3. Withdraw money") # print the option to withdraw money
    print("4. Exit") # print the option to exit

    choice=int(input("Enter your choice between 1 to 4")) #ask the user to input their choice and convert it to an integer

    if choice==1: # check if the user has selected option 1
        print(f"Your current Bank Balance is {balance}") # print the current balance
    elif choice==2: # check if the user has selected option 2
        amount=float(input("Enter amount you would like to deposit"))
        if amount>0: # check if the amount entered is greater than 0
            balance=balance+amount # add the amount to the balance
            print(f"succesfully Deposited {amount}")
            print(f"Latest balance is {balance}")
        else: # check if the amount entered is less than or equal to 0
            print("Invalid amount , Please provide latest amount to Deposit")
    elif choice==3: # check if the user has selected option 3
        amount=float(input("Enter amount you would like to withdraw"))
        if amount>0 and amount<=balance: # check if the amount entered is greater than 0 and less than or equal to the balance
            balance=balance-amount # subtract the amount from the balance
            print(f"succesfully withdraw {amount}")
            print(f"Latest balance is {balance}")
        elif amount>balance: # check if the amount entered is greater than the balance
            print(f"Insufficient balance, Your current balance is {balance}")
        else: # check if the amount entered is less than or equal to 0
            print("Invalid amount , Please provide latest amount to Deposit")
    elif choice==4: # check if the user has selected option 4
        print("Thank you for visiting ATM")
        is_running=False # set is_running to False to exit the loop
    else:
        print("Invalid choice, Select any choice between 1 to 4")
```
