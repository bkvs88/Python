### Calculate factorial using while

```python
num_inp = int(input("enter any number for factorial"))
fact=1 ## initializing fact variable to 1
temp=num_inp ## assigning the input number to temp variable for later use in print statement
while num_inp >= 1 : #looping through the input number until it is greater than or equal to 1
    fact = fact * num_inp ## multiplying the fact variable with the input number and assigning it back to fact variable
    num_inp = num_inp - 1 ## decrementing the input number by 1 in each iteration

print(f"Total factorial of {temp} is {fact}") # printing the final factorial value of the input number using f-string formatting
```
