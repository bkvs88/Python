### Given a list containing duplicate elements, create another list containing only unique elements without using set().

```python
numbers_list= [0, 15, 30, 15, 45,45, 60, 75, 90,90, 105, 120, 135, 150, 165, 180, 195] #list of numbers to remove duplicates from
duplicate_list=[] # initializing an empty list to store unique numbers
for i in numbers_list: # looping through each number in the list
    if i not in duplicate_list: # checking if the current number is not already in the duplicate_list
        duplicate_list.append(i) # adding the number to the duplicate_list if it is not already present
print(duplicate_list) # printing the list of unique numbers after removing duplicates
```
