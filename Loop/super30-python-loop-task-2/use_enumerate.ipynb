### Given names = ["Aman", "Ravi", "Sudhanshu", "Priya", "Anjali"] use enumerate() to display

```python
names = ["Aman", "Ravi", "Sudhanshu", "Priya", "Anjali"] ## initializing a list of names
for index,name in enumerate(names,start=1): ## looping through the list of names using enumerate to get index and name, starting index from 1
    print(index,name) ## printing the index and name in each iteration
```
