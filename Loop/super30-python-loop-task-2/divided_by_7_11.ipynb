#Iterate from 1–100 and stop when you encounter the first number divisible by both 7 and 11.

```python
numbers = range(1,101) # looping through numbers from 1 to 100 using range function
for i in numbers: # looping through numbers from 1 to 100 using for loop
    if i % 7 == 0 and i % 11 == 0: # checking if the number is divisible by both 7 and 11
        print(f"Number {i} found divisible by 7 and 11")
        break
```
