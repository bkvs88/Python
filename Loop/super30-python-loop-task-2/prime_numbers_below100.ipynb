### Print all prime numbers between 1 and 100.

```python
prime_list=[] # initializing an empty list to store prime numbers
for i in range(2,101): # looping through numbers from 2 to 100 using range function
    for number in range(2,i): # looping through numbers from 2 to the current number (exclusive) to check for factors
        if i%number == 0: # checking if the current number is divisible by any number in the inner loop (i.e., if it has a factor other than 1 and itself)
            break #breaking the inner loop if a factor is found, indicating that the number is not prime
    else:
        print(i,end=" ") # printing the prime number if no factors are found, indicating that the number is prime, and using end=" " to print on the same line
```
