### Search for a user-provided number inside a list. Use for-else to print "Number Found" "Number Not Found"

```python
#Search for a user-provided number inside a list. Use for-else to print:
numbers = [12, 7, 9, 20, 33, 42, 8, 15, 25, 99, 86] # list of numbers to search from
number_to_search=int(input("enter number to search")) # taking user input for the number to search
for i in numbers: # looping through the list of numbers
    if number_to_search == i: # checking if the user-provided number is present in the list
        print (f"Number Found") # printing the number if found
        break # breaking the loop if the number is found
else:
    print("Number Not Found")
```
