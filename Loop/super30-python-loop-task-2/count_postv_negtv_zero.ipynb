### Given numbers = [10, -4, 8, -2, 0, 15, -9, 21] count positive numbers , negative numbers,  zeros

```python
numbers_list= numbers = [10, -4, 8, -2, 0, 15, -9, 21]#list of numbers to count positive, negative and zero numbers
zero_count=0 # initializing a variable to count the number of zeros in the list
positive_count=0 # initializing a variable to count the number of positive numbers in the list
negative_count=0 # initializing a variable to count the number of negative numbers in the list
for i in numbers_list: # looping through each number in the list
    if i == 0: # checking if the current number is zero
        zero_count=zero_count+1 # incrementing the zero_count variable by 1 if the current number is zero
    elif i > 0 : # checking if the current number is positive
        positive_count=positive_count+1 # incrementing the positive_count variable by 1 if the current number is positive
    else: # checking if the current number is negative
        negative_count=negative_count+1 # incrementing the negative_count variable by 1 if the current number is negative
print(f"positive numbers count is : {positive_count}\nnegative numbers count is : {negative_count}\nzeros count is : {zero_count}")
```
