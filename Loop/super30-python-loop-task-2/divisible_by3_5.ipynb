### Find all numbers between 1 and 200 divisible by both 3 and 5.

```python
numbers=range(1,201) #looping through numbers from 0 to 200 using range function
Number_div_by_3_5=[] # initializing an empty list to store numbers divisible by both 3 and 5
for i in numbers: # looping through each number in the range
    if i%3 ==0 and i%5==0: # checking if the current number is divisible by both 3 and 5
        Number_div_by_3_5.append(i) # adding the number to the list if it is divisible by both 3 and 5
print(Number_div_by_3_5) # printing the list of numbers that are divisible by both 3 and 5
```
