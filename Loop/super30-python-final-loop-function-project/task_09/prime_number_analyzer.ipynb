Prime Number Analyzer
Take two numbers representing a range.
Create functions to
find prime numbers
count primes
calculate their sum
display the largest prime found



````python
def find_primes(start, end):
    """Return all prime numbers in the inclusive range from start to end."""
    # Create an empty list to store the prime numbers.
    primes = []

    # Check every number in the requested range.
    for number in range(start, end + 1):
        # Ignore numbers smaller than 2 because they are not prime.
        if number < 2:
            continue

        # Assume the current number is prime.
        is_prime = True

        # Test divisors up to the square root of the number.
        for divisor in range(2, int(number ** 0.5) + 1):
            # If evenly divisible, the number is not prime.
            if number % divisor == 0:
                is_prime = False
                break

        # Add the number if no divisor was found.
        if is_prime:
            primes.append(number)

    # Return the list of prime numbers.
    return primes


def count_primes(primes):
    """Return the number of prime numbers in a list."""
    # Return the length of the prime numbers list.
    return len(primes)


def sum_primes(primes):
    """Return the sum of all prime numbers in a list."""
    # Return the total of all prime numbers.
    return sum(primes)


def display_largest_prime(primes):
    """Display and return the largest prime number, if one exists."""
    # Check whether the list contains any prime numbers.
    if primes:
        # Find the largest prime number.
        largest = max(primes)

        # Display the largest prime number.
        print(f"Largest prime: {largest}")

        # Return the largest prime number.
        return largest

    # Display a message when no primes are found.
    print("No prime numbers were found.")

    # Return None because there is no largest prime.
    return None


# Ask the user for the beginning of the range.
start = int(input("Enter the start of the range: "))

# Ask the user for the end of the range.
end = int(input("Enter the end of the range: "))

# Swap the values if the user enters them in reverse order.
if start > end:
    start, end = end, start

# Find all prime numbers in the range.
primes = find_primes(start, end)

# Display the prime numbers.
print(f"Prime numbers: {primes}")

# Count and display the number of primes.
print(f"Number of primes: {count_primes(primes)}")

# Calculate and display the sum of the primes.
print(f"Sum of primes: {sum_primes(primes)}")

# Display the largest prime.
display_largest_prime(primes)
````
