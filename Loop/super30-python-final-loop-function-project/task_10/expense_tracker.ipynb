Expense Tracker

Allow a user to repeatedly enter

expense name

amount

Provide options to

add expense

view expenses

calculate total

find highest expense

exit

````python
# Store each expense as a dictionary inside this list.
expenses = []


def add_expense():
    """Prompt the user for an expense name and amount, then save it."""
    # Ask the user to enter the expense name.
    name = input("Enter expense name: ").strip()

    # Check that the expense name is not empty.
    if not name:
        print("Expense name cannot be empty.")
        return

    # Repeatedly ask for the amount until a valid value is entered.
    while True:
        try:
            # Convert the entered amount to a floating-point number.
            amount = float(input("Enter expense amount: "))

            # Ensure the amount is not negative.
            if amount < 0:
                print("Amount cannot be negative.")
                continue

            # Stop asking once a valid amount is entered.
            break
        except ValueError:
            # Display an error when the input is not a number.
            print("Please enter a valid number.")

    # Add the expense details to the expenses list.
    expenses.append({"name": name, "amount": amount})

    # Confirm that the expense was added.
    print("Expense added successfully.")


def view_expenses():
    """Display all recorded expenses."""
    # Check whether the expenses list is empty.
    if not expenses:
        print("No expenses recorded.")
        return

    # Display a heading for the expense list.
    print("\nExpenses:")

    # Loop through each expense with its position number.
    for number, expense in enumerate(expenses, start=1):
        # Display the expense name and amount.
        print(f"{number}. {expense['name']}: ${expense['amount']:.2f}")


def calculate_total():
    """Calculate and display the total of all expenses."""
    # Add all expense amounts together.
    total = sum(expense["amount"] for expense in expenses)

    # Display the total amount.
    print(f"Total expenses: ${total:.2f}")


def find_highest_expense():
    """Find and display the expense with the highest amount."""
    # Check whether the expenses list is empty.
    if not expenses:
        print("No expenses recorded.")
        return

    # Find the expense with the greatest amount.
    highest = max(expenses, key=lambda expense: expense["amount"])

    # Display the highest expense.
    print(f"Highest expense: {highest['name']} - ${highest['amount']:.2f}")


def display_menu():
    """Display the available expense tracker options."""
    # Print each menu option.
    print("\nExpense Tracker")
    print("1. Add expense")
    print("2. View expenses")
    print("3. Calculate total")
    print("4. Find highest expense")
    print("5. Exit")


# Continue running the program until the user chooses to exit.
while True:
    # Display the menu options.
    display_menu()

    # Ask the user to select an option.
    choice = input("Choose an option: ").strip()

    # Run the function matching the selected option.
    if choice == "1":
        add_expense()
    elif choice == "2":
        view_expenses()
    elif choice == "3":
        calculate_total()
    elif choice == "4":
        find_highest_expense()
    elif choice == "5":
        print("Goodbye!")
        break
    else:
        # Handle an invalid menu selection.
        print("Invalid option. Please try again.")
````
