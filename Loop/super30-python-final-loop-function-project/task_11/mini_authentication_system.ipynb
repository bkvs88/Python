Mini Authentication System

Create a small application supporting

predefined username/password

maximum login attempts

successful login

failed login

logout

retry logic

Use functions and loops appropriately.



````python
from getpass import getpass  # Import getpass to hide password input.


# Store the predefined username and password.
PREDEFINED_USERNAME = "admin"
PREDEFINED_PASSWORD = "password123"

# Define the maximum number of login attempts.
MAX_LOGIN_ATTEMPTS = 3


def authenticate(username, password):
    """Return True when the provided credentials are valid."""
    # Compare the entered credentials with the predefined credentials.
    return (
        username == PREDEFINED_USERNAME
        and password == PREDEFINED_PASSWORD
    )


def login():
    """Allow the user to retry login until success or attempts are exhausted."""
    # Repeat the login process up to the maximum number of attempts.
    for attempt in range(1, MAX_LOGIN_ATTEMPTS + 1):
        # Ask the user to enter a username.
        username = input("Enter username: ")

        # Ask the user to enter a hidden password.
        password = getpass("Enter password: ")

        # Check whether the entered credentials are correct.
        if authenticate(username, password):
            # Display a success message when authentication succeeds.
            print("Login successful.")
            return True

        # Calculate how many attempts remain.
        remaining_attempts = MAX_LOGIN_ATTEMPTS - attempt

        # Display a failed-login message.
        print("Login failed.")

        # Inform the user about remaining attempts.
        if remaining_attempts > 0:
            print(f"Attempts remaining: {remaining_attempts}")
        else:
            print("Maximum login attempts reached.")

    # Return False when all attempts fail.
    return False


def logout():
    """Log the user out of the authentication system."""
    # Display a logout confirmation message.
    print("You have been logged out.")


def main():
    """Run the mini authentication system."""
    # Display the application title.
    print("Mini Authentication System")

    # Attempt to log the user in.
    logged_in = login()

    # Continue only when login is successful.
    if logged_in:
        # Display the available action.
        print("You are now logged in.")

        # Ask whether the user wants to log out.
        action = input("Enter 'logout' to log out: ").strip().lower()

        # Log out when the requested action is entered.
        if action == "logout":
            logout()
        else:
            # Handle an unknown action.
            print("Invalid action.")
    else:
        # Display a message when login fails completely.
        print("Access denied.")


# Start the application.
main()
````
