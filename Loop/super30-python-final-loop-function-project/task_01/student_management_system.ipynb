````python
def accept_student_marks():
    """Collect and validate one student's name and subject marks."""
    try:
        student_name = input("Enter student name: ").strip()
        if not student_name:
            raise ValueError("Student name cannot be empty.")

        subject_count = int(input("Enter number of subjects: "))
        if subject_count <= 0:
            raise ValueError("Number of subjects must be greater than zero.")

        marks = []
        for subject_number in range(1, subject_count + 1):
            mark = float(input(f"Enter marks for subject {subject_number}: "))
            if not 0 <= mark <= 100:
                raise ValueError(
                    f"Marks for subject {subject_number} must be between 0 and 100."
                )
            marks.append(mark)

        return student_name, marks
    except ValueError as error:
        print(f"Input error: {error}")
        return None, None


def calculate_total(marks):
    """Return the total of a collection of numeric marks."""
    if not marks:
        raise ValueError("At least one mark is required.")
    if not all(isinstance(mark, (int, float)) and not isinstance(mark, bool)
               for mark in marks):
        raise TypeError("Marks must contain only integer or float values.")

    # Sum all subject marks to calculate the student's total.
    return sum(marks)


def calculate_percentage(total, subject_count):
    """Calculate the average percentage from total marks and subject count."""
    if subject_count <= 0:
        raise ValueError("Subject count must be greater than zero.")

    # Each subject is assumed to be marked out of 100.
    return total / subject_count


def assign_grade(percentage):
    """Return a letter grade based on the student's percentage."""
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"


def find_failed_subjects(marks):
    """Return the subject number and mark for each subject below 40."""
    # Keep the subject number so the student can identify the failed subject.
    return [
        (subject_number, mark)
        for subject_number, mark in enumerate(marks, start=1)
        if mark < 40
    ]


def determine_pass_fail(marks):
    """Return Pass when every subject mark is at least 40."""
    if not marks:
        raise ValueError("At least one mark is required.")
    return "Pass" if not find_failed_subjects(marks) else "Fail"


def display_result(student_name, marks):
    """Calculate and display a student's result and any failure reasons."""
    total_marks = calculate_total(marks)
    percentage = calculate_percentage(total_marks, len(marks))
    grade = assign_grade(percentage)
    result_status = determine_pass_fail(marks)
    failed_subjects = find_failed_subjects(marks)

    print("\n--- Student Result ---")
    print(f"Name: {student_name}")
    print(f"Total Marks: {total_marks:g}")
    print(f"Percentage: {percentage:.2f}%")
    print(f"Grade: {grade}")
    print(f"Result: {result_status}")

    if failed_subjects:
        print("Failure reason(s):")
        for subject_number, mark in failed_subjects:
            print(
                f"Subject {subject_number} scored {mark:g}, "
                "which is below the passing mark of 40."
            )


def main():
    """Run the student marks collection and result display workflow."""
    try:
        student_name, student_marks = accept_student_marks()
        if student_name is not None and student_marks is not None:
            display_result(student_name, student_marks)
    except (EOFError, KeyboardInterrupt):
        print("\nApplication closed safely.")
    except (TypeError, ValueError) as error:
        print(f"Unable to calculate the result: {error}")
````
