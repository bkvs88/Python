
Employee Salary Analyzer

Given employee salaries, create functions to determine

total payroll

average salary

highest salary

lowest salary

employees earning above average

````python
# Store employee names as keys and their salaries as values.
employee_salaries = {
    "Alice": 72000,
    "Bob": 55000,
    "Charlie": 68000,
    "Diana": 81000,
}


def total_payroll(salaries):
    """Return the total of all employee salaries."""
    # Add all salary values together.
    return sum(salaries.values())


def average_salary(salaries):
    """Return the average employee salary."""
    # Raise an error if no salaries were provided.
    if not salaries:
        raise ValueError("The salary collection cannot be empty.")

    # Divide the total payroll by the number of employees.
    return total_payroll(salaries) / len(salaries)


def highest_salary(salaries):
    """Return the employee with the highest salary."""
    # Raise an error if no salaries were provided.
    if not salaries:
        raise ValueError("The salary collection cannot be empty.")

    # Find the employee whose salary is highest.
    employee = max(salaries, key=salaries.get)

    # Return the employee name and salary.
    return employee, salaries[employee]


def lowest_salary(salaries):
    """Return the employee with the lowest salary."""
    # Raise an error if no salaries were provided.
    if not salaries:
        raise ValueError("The salary collection cannot be empty.")

    # Find the employee whose salary is lowest.
    employee = min(salaries, key=salaries.get)

    # Return the employee name and salary.
    return employee, salaries[employee]


def employees_above_average(salaries):
    """Return employees whose salaries are above the average salary."""
    # Calculate the average salary.
    average = average_salary(salaries)

    # Select employees whose salaries exceed the average.
    return {
        employee: salary
        for employee, salary in salaries.items()
        if salary > average
    }


# Display the total payroll.
print(f"Total payroll: ${total_payroll(employee_salaries):,.2f}")

# Display the average salary.
print(f"Average salary: ${average_salary(employee_salaries):,.2f}")

# Display the highest-paid employee and salary.
highest_employee, highest_amount = highest_salary(employee_salaries)
print(f"Highest salary: {highest_employee} (${highest_amount:,.2f})")

# Display the lowest-paid employee and salary.
lowest_employee, lowest_amount = lowest_salary(employee_salaries)
print(f"Lowest salary: {lowest_employee} (${lowest_amount:,.2f})")

# Display employees earning above the average salary.
print("Employees earning above average:")
for employee, salary in employees_above_average(employee_salaries).items():
    print(f"- {employee}: ${salary:,.2f}")
````
