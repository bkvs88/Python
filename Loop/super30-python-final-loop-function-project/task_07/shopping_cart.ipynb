Shopping Cart

Create a simple cart where users can

add products

remove products

view cart

calculate bill

exit

The application should continue until Exit is selected.

````python

# Store available products and their prices.
PRODUCTS = {
    "apple": 1.50,
    "bread": 2.00,
    "milk": 2.75,
    "eggs": 3.50,
    "rice": 4.25,
}

# Store products added to the shopping cart.
cart = {}


def display_products():
    """Display all products available for purchase."""
    print("\nAvailable Products:")
    for product, price in PRODUCTS.items():
        print(f"- {product.title()}: ${price:.2f}")


def add_product():
    """Add a selected product and quantity to the cart."""
    display_products()
    product = input("\nEnter the product to add: ").strip().lower()

    if product not in PRODUCTS:
        print("Product not found.")
        return

    try:
        quantity = int(input("Enter the quantity: "))
        if quantity <= 0:
            print("Quantity must be greater than zero.")
            return
    except ValueError:
        print("Please enter a valid whole number.")
        return

    cart[product] = cart.get(product, 0) + quantity
    print(f"{quantity} {product}(s) added to the cart.")


def remove_product():
    """Remove a selected product or quantity from the cart."""
    if not cart:
        print("\nYour cart is empty.")
        return

    view_cart()
    product = input("\nEnter the product to remove: ").strip().lower()

    if product not in cart:
        print("That product is not in your cart.")
        return

    try:
        quantity = int(input("Enter the quantity to remove: "))
        if quantity <= 0:
            print("Quantity must be greater than zero.")
            return
    except ValueError:
        print("Please enter a valid whole number.")
        return

    if quantity >= cart[product]:
        del cart[product]
        print(f"{product.title()} removed from the cart.")
    else:
        cart[product] -= quantity
        print(f"{quantity} {product}(s) removed from the cart.")


def view_cart():
    """Display the products currently in the cart."""
    if not cart:
        print("\nYour cart is empty.")
        return

    print("\nShopping Cart:")
    for product, quantity in cart.items():
        price = PRODUCTS[product]
        subtotal = price * quantity
        print(f"- {product.title()}: {quantity} x ${price:.2f} = ${subtotal:.2f}")


def calculate_bill():
    """Calculate and display the total cost of all cart items."""
    if not cart:
        print("\nYour cart is empty.")
        return

    total = sum(PRODUCTS[product] * quantity for product, quantity in cart.items())

    print("\nBill:")
    view_cart()
    print(f"Total: ${total:.2f}")


def main():
    """Run the shopping cart application until the user exits."""
    while True:
        print("\nShopping Cart Menu")
        print("1. Add product")
        print("2. Remove product")
        print("3. View cart")
        print("4. Calculate bill")
        print("5. Exit")

        choice = input("Enter your choice: ").strip()

        if choice == "1":
            add_product()
        elif choice == "2":
            remove_product()
        elif choice == "3":
            view_cart()
        elif choice == "4":
            calculate_bill()
        elif choice == "5":
            print("Thank you for shopping!")
            break
        else:
            print("Invalid choice. Please try again.")


# Start the shopping cart application.
main()
````
