````python
def display_menu():
    """Display the banking application menu."""
    print("\n--- Banking Application ---")
    print("1. Check balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Transaction history")
    print("5. Exit")


def deposit(current_balance, transaction_history):
    """Deposit a positive amount and return the updated balance."""
    try:
        deposit_amount = float(input("Enter deposit amount: "))
        if deposit_amount <= 0:
            print("Amount must be greater than zero.")
            return current_balance

        # Record the transaction only after the balance is updated successfully.
        updated_balance = current_balance + deposit_amount
        transaction_history.append(f"Deposited: ${deposit_amount:.2f}")
        print(f"${deposit_amount:.2f} deposited successfully.")
        return updated_balance
    except ValueError:
        print("Please enter a valid amount.")
        return current_balance


def withdraw(current_balance, transaction_history):
    """Withdraw a valid amount when sufficient funds are available."""
    try:
        withdrawal_amount = float(input("Enter withdrawal amount: "))
        if withdrawal_amount <= 0:
            print("Amount must be greater than zero.")
        elif withdrawal_amount > current_balance:
            print("Insufficient funds.")
        else:
            # Record the withdrawal only after validating the available balance.
            updated_balance = current_balance - withdrawal_amount
            transaction_history.append(f"Withdrew: ${withdrawal_amount:.2f}")
            print(f"${withdrawal_amount:.2f} withdrawn successfully.")
            return updated_balance
    except ValueError:
        print("Please enter a valid amount.")

    return current_balance


def show_history(transaction_history):
    """Display all completed account transactions."""
    if transaction_history:
        print("\nTransaction history:")
        for transaction in transaction_history:
            print(transaction)
    else:
        print("No transactions yet.")


def main():
    """Run the banking application until the user chooses to exit."""
    account_balance = 0.0
    transaction_history = []

    while True:
        display_menu()

        try:
            menu_choice = input("Enter your choice: ")
        except (EOFError, KeyboardInterrupt):
            print("\nApplication closed safely.")
            break

        # Route each menu choice to the matching banking operation.
        if menu_choice == "1":
            print(f"Current balance: ${account_balance:.2f}")
        elif menu_choice == "2":
            account_balance = deposit(account_balance, transaction_history)
        elif menu_choice == "3":
            account_balance = withdraw(account_balance, transaction_history)
        elif menu_choice == "4":
            show_history(transaction_history)
        elif menu_choice == "5":
            print("Thank you for using the banking application.")
            break
        else:
            print("Invalid choice. Please try again.")


# Start the banking application.
main()
````
