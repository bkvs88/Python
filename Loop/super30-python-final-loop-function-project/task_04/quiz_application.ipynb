
Quiz Application

Create at least 5 Python questions.

The application should

display one question at a time

accept answers

check answers

maintain score

show final percentage

````python
# Store the quiz questions and their correct answers.
questions = [
    {
        # Define the first question.
        "question": "What keyword defines a function in Python?",
        # Store the correct answer.
        "answer": "def"
    },
    {
        # Define the second question.
        "question": "Which data type stores key-value pairs?",
        # Store the correct answer.
        "answer": "dictionary"
    },
    {
        # Define the third question.
        "question": "What symbol starts a comment in Python?",
        # Store the correct answer.
        "answer": "#"
    },
    {
        # Define the fourth question.
        "question": "Which function displays output?",
        # Store the correct answer.
        "answer": "print"
    },
    {
        # Define the fifth question.
        "question": "What type of loop iterates over a sequence?",
        # Store the correct answer.
        "answer": "for"
    }
]


def display_question(number, total, item):
    """Display a single quiz question with its number."""
    # Show which question is currently being asked.
    print(f"Question {number} of {total}")
    # Show the question text.
    print(item["question"])


def check_answer(user_answer, correct_answer):
    """Return True when the user's answer matches the correct answer."""
    # Compare the normalized answers, ignoring letter case.
    return user_answer.strip().lower() == correct_answer


def run_quiz():
    """Run the quiz, accepting answers, and return the final score."""
    # Initialize the user's score.
    score = 0

    # Loop through each question with a one-based number.
    for number, item in enumerate(questions, start=1):
        # Display the current question.
        display_question(number, len(questions), item)

        try:
            # Read and normalize the user's answer.
            user_answer = input("Your answer: ").strip().lower()
        except (EOFError, KeyboardInterrupt):
            # Stop the quiz if the user interrupts input.
            print("\nQuiz cancelled.")
            return 0

        # Check whether the answer is correct.
        if check_answer(user_answer, item["answer"]):
            # Display a success message.
            print("Correct!\n")
            # Increase the score by one.
            score += 1
        else:
            # Display the correct answer.
            print(f"Incorrect. The correct answer is: {item['answer']}\n")

    # Return the total number of correct answers.
    return score


def show_final_result(score):
    """Display the final score and percentage."""
    # Calculate the final percentage.
    percentage = score / len(questions) * 100

    # Display the completion message and results.
    print("Quiz complete!")
    print(f"Score: {score}/{len(questions)}")
    print(f"Percentage: {percentage:.1f}%")


def main():
    """Run the complete quiz application."""
    # Display the quiz title.
    print("Python Quiz\n")

    # Run the quiz and capture the final score.
    final_score = run_quiz()

    # Display the final result once all questions are answered.
    show_final_result(final_score)


# Start the quiz application.
main()
````
