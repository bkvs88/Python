Number Analysis Tool

Create a function that accepts a list and returns

largest number

smallest number

total

average

even count

odd count

positive count

negative count

Do not use min(), max(), or sum().

````python
def analyze_numbers(numbers):
    """Return basic statistics for a list of numbers without using min(), max(), or sum()."""

    # Reject an empty list because statistics cannot be calculated for it.
    if not numbers:
        raise ValueError("The list must contain at least one number.")

    # Initialize all calculated values using the first number.
    largest = numbers[0]
    smallest = numbers[0]
    total = 0
    even_count = 0
    odd_count = 0
    positive_count = 0
    negative_count = 0

    # Process each number in the list.
    for number in numbers:
        # Update the largest number when a larger value is found.
        if number > largest:
            largest = number

        # Update the smallest number when a smaller value is found.
        if number < smallest:
            smallest = number

        # Add the current number to the running total.
        total += number

        # Count even and odd numbers.
        if number % 2 == 0:
            even_count += 1
        else:
            odd_count += 1

        # Count positive and negative numbers.
        if number > 0:
            positive_count += 1
        elif number < 0:
            negative_count += 1

    # Calculate the average using the list length.
    average = total / len(numbers)

    # Return all results in a dictionary.
    return {
        "largest": largest,
        "smallest": smallest,
        "total": total,
        "average": average,
        "even_count": even_count,
        "odd_count": odd_count,
        "positive_count": positive_count,
        "negative_count": negative_count,
    }


# Create a sample list of numbers.
numbers = [10, -3, 7, 0, 4, -8]

# Analyze the numbers and store the results.
results = analyze_numbers(numbers)

# Display the results.
print(results)
````
