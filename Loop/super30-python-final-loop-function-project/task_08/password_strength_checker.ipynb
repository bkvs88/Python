Password Strength Checker

Create a function that checks whether a password contains

uppercase

lowercase

number

special character

minimum 8 characters

Return a meaningful strength/result.

````python
def check_password_strength(password):
    """
    Check a password for uppercase, lowercase, number, special character,
    and minimum length requirements.

    Args:
        password (str): The password to evaluate.

    Returns:
        dict: A result containing individual checks and overall strength.
    """
    # Ensure the input is treated as a string.
    password = str(password)

    # Check whether the password contains at least one uppercase letter.
    has_uppercase = any(character.isupper() for character in password)

    # Check whether the password contains at least one lowercase letter.
    has_lowercase = any(character.islower() for character in password)

    # Check whether the password contains at least one number.
    has_number = any(character.isdigit() for character in password)

    # Check whether the password contains at least one non-alphanumeric character.
    has_special_character = any(
        not character.isalnum() for character in password
    )

    # Check whether the password contains at least 8 characters.
    has_minimum_length = len(password) >= 8

    # Store all password requirements and their results.
    requirements = {
        "uppercase": has_uppercase,
        "lowercase": has_lowercase,
        "number": has_number,
        "special_character": has_special_character,
        "minimum_length": has_minimum_length,
    }

    # Count how many requirements the password satisfies.
    passed_requirements = sum(requirements.values())

    # Assign an overall strength based on the number of satisfied requirements.
    if passed_requirements == 5:
        strength = "Strong"
    elif passed_requirements >= 3:
        strength = "Moderate"
    else:
        strength = "Weak"

    # Return the strength and detailed requirement results.
    return {
        "strength": strength,
        "requirements": requirements,
    }


# Ask the user to enter a password.
password = input("Enter a password: ")

# Check the password strength.
result = check_password_strength(password)

# Display the overall password strength.
print(f"\nPassword strength: {result['strength']}")

# Display the result of each individual requirement.
for requirement, passed in result["requirements"].items():
    # Convert the Boolean result into a readable status.
    status = "Passed" if passed else "Failed"

    # Print the requirement status.
    print(f"{requirement.replace('_', ' ').title()}: {status}")
````
