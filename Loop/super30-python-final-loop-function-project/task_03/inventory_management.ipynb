Inventory Management

Maintain products containing

product name

price

quantity

Provide functions to

add product

display products

search product

update quantity

calculate total inventory value

````python
inventory = {}


def add_product(name, price, quantity):
    """Add a new product to the inventory.

    Args:
        name (str): The name of the product.
        price (float): The price of the product.
        quantity (int): The available quantity of the product.
    """
    # Reject a product name that already exists in the inventory.
    if name in inventory:
        print("Product already exists.")
        return

    # Ensure the price and quantity are not negative.
    if price < 0 or quantity < 0:
        print("Price and quantity must be non-negative.")
        return

    # Store the new product with its price and quantity.
    inventory[name] = {
        "price": price,
        "quantity": quantity
    }


def display_products():
    """Display all products currently stored in the inventory."""
    # Show a message when the inventory is empty.
    if not inventory:
        print("Inventory is empty.")
        return

    # Print every product with its price and quantity.
    for name, details in inventory.items():
        print(
            f"Product: {name}, "
            f"Price: ${details['price']:.2f}, "
            f"Quantity: {details['quantity']}"
        )


def search_product(name):
    """Find and display a product by its name.

    Args:
        name (str): The name of the product to search for.

    Returns:
        dict: The product details if found; otherwise, None.
    """
    # Retrieve the product from inventory without raising an error if missing.
    product = inventory.get(name)

    # Check whether the product was found.
    if product:
        # Display the product's name, price, and quantity.
        print(
            f"Product: {name}, "
            f"Price: ${product['price']:.2f}, "
            f"Quantity: {product['quantity']}"
        )
        # Return the product details.
        return product

    # Display a message when the product is not found.
    print("Product not found.")
    # Return None to indicate that no product was found.
    return None


def update_quantity(name, quantity):
    """Update the quantity of an existing product.

    Args:
        name (str): The name of the product to update.
        quantity (int): The new quantity to assign.
    """
    # Reject updates for products that do not exist.
    if name not in inventory:
        print("Product not found.")
        return

    # Ensure the new quantity is not negative.
    if quantity < 0:
        print("Quantity must be non-negative.")
        return

    # Assign the new quantity to the product.
    inventory[name]["quantity"] = quantity


def calculate_total_inventory_value():
    """Return the total value of all products in the inventory."""
    # Multiply each product's price by its quantity and sum the results.
    return sum(
        product["price"] * product["quantity"]
        for product in inventory.values()
    )
````

````python
# Example usage
add_product("Laptop", 999.99, 5)
add_product("Mouse", 25.50, 10)

display_products()
search_product("Laptop")
update_quantity("Mouse", 15)

print(f"Total inventory value: ${calculate_total_inventory_value():.2f}")
````
