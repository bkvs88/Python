Super30 Python Utility Application

Create your own menu-driven application containing at least five utilities.

````python
"""Super30 Python Utility Application.

A menu-driven program containing multiple useful utilities.
"""


def calculator():
    """Perform a basic arithmetic calculation."""
    print("\n--- Calculator ---")

    try:
        first_number = float(input("Enter the first number: "))
        operator = input("Enter an operator (+, -, *, /): ")
        second_number = float(input("Enter the second number: "))

        # Select the arithmetic operation requested by the user.
        if operator == "+":
            result = first_number + second_number
        elif operator == "-":
            result = first_number - second_number
        elif operator == "*":
            result = first_number * second_number
        elif operator == "/":
            result = first_number / second_number
        else:
            print("Invalid operator.")
            return

        print(f"Result for {first_number}{operator}{second_number}: {result}")
    except ValueError:
        print("Error: Please enter valid numbers.")
    except ZeroDivisionError:
        print("Error: Division by zero is not allowed.")


def check_palindrome():
    """Check whether a word or phrase is a palindrome."""
    text = input("\nEnter a word or phrase: ")

    # Ignore spaces, punctuation, and letter case for the comparison.
    cleaned_text = "".join(character.lower() for character in text if character.isalnum())

    if cleaned_text == cleaned_text[::-1]:
        print(f'The text "{text}" is a palindrome.')
    else:
        print(f'The text "{text}" is not a palindrome.')


def check_prime():
    """Check whether a number is prime."""
    try:
        number = int(input("\nEnter an integer: "))
    except ValueError:
        print("Error: Please enter a valid integer.")
        return

    if number < 2:
        print(f"The number {number} is not prime.")
        return

    # A divisor only needs to be searched for up to the square root.
    is_prime = True
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            is_prime = False
            break

    if is_prime:
        print(f"The number {number} is prime.")
    else:
        print(f"The number {number} is not prime.")


def count_words():
    """Count the number of words in a sentence."""
    sentence = input("\nEnter a sentence: ")

    # split() handles multiple spaces and ignores leading or trailing spaces.
    words = sentence.split()
    print(f'Word count for "{sentence}": {len(words)}')


def convert_temperature():
    """Convert temperature between Celsius and Fahrenheit."""
    print("\n1. Celsius to Fahrenheit")
    print("2. Fahrenheit to Celsius")

    choice = input("Choose an option: ")

    try:
        temperature = float(input("Enter the temperature: "))
    except ValueError:
        print("Error: Please enter a valid temperature.")
        return

    # Apply the formula selected by the user.
    if choice == "1":
        converted_temperature = (temperature * 9 / 5) + 32
        print(f"Temperature {temperature} in Fahrenheit: {converted_temperature:.2f}")
    elif choice == "2":
        converted_temperature = (temperature - 32) * 5 / 9
        print(f"Temperature {temperature} in Celsius: {converted_temperature:.2f}")
    else:
        print("Invalid choice.")


def main():
    """Display the menu and run the selected utility."""
    while True:
        print("\n===== Super30 Python Utility Application =====")
        print("1. Calculator")
        print("2. Palindrome Checker")
        print("3. Prime Number Checker")
        print("4. Word Counter")
        print("5. Temperature Converter")
        print("6. Exit")

        try:
            choice = input("Enter your choice: ")
        except (EOFError, KeyboardInterrupt):
            print("\nApplication closed safely.")
            break

        # Call the utility that matches the selected menu option.
        if choice == "1":
            calculator()
        elif choice == "2":
            check_palindrome()
        elif choice == "3":
            check_prime()
        elif choice == "4":
            count_words()
        elif choice == "5":
            convert_temperature()
        elif choice == "6":
            print("Thank you for using the Super30 Python Utility Application.")
            break
        else:
            print("Invalid choice. Please select an option from 1 to 6.")
````
