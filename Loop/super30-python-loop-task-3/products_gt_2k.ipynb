### print only products costing more than ₹2,000.

```python
products = {"Laptop": 55000,"Phone": 30000,"Headphones": 2000,"Mouse": 700,"Keyboard": 1500} #initializing a dictionary with product names as keys and their prices as values
for i in products.keys(): # looping through each product name in the dictionary
    if products[i] > 2000: # checking if the price of the current product is greater than 2000
        print(i ,end= " ") # printing the product name if its price is greater than 2000 and using end=" " to print on the same line
```
