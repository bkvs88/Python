### Create this number pattern
### 1
### 12
### 123
### 1234
### 12345

```python
j='' #initializing an empty string variable to store the concatenated values
for i in range(1,6): # looping through the range of numbers from 1 to 5
    j = f"{j}{i}" #concatenating the current number to the string variable j using f-string formatting
    print(j)
```
