### From the same list, find the highest and lowest transaction without max() and min().

```python
transactions = [1200, 450, 800, 1500, 2300, 700, 100] #list of transaction values
sum=0 #initializing a variable to store the sum of transaction values
highestval=transactions[0] #initializing a variable to store the highest transaction value
lowestval=transactions[0] #initializing a variable to store the lowest transaction value
for i in transactions: # looping through each transaction value in the list
    sum=sum+i # adding the current transaction value to the sum
    if i>highestval: # checking if the current transaction value is greater than the highest transaction value
        highestval=i # updating the highest transaction value if the current transaction value is greater
    if i<lowestval: # checking if the current transaction value is less than the lowest transaction value
        lowestval=i #
print(f"Total Transaction value is {sum}")
print(f"Max Transaction value is {highestval}")
print(f"Low Transaction value is {lowestval}")
```
