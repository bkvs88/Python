```python
#count how many students scored 90 +75–89 50–74 below 50
marks = [78, 92, 45, 67, 88, 53, 99]
count_90_plus=0
count_75_to_89=0
count_50_to_74=0
count_below_50=0
for i in marks:
    if i >= 90:
        count_90_plus=count_90_plus+1
    elif i < 90 and i>=75:
        count_75_to_89=count_75_to_89+1
    elif i<75 and i>=50:
        count_50_to_74=count_50_to_74+1
    elif i<50:
        count_below_50=count_below_50+1

print(f"{count_90_plus} students scored 90+")
print(f"{count_75_to_89} students scored between 75 to 89")
print(f"{count_50_to_74} students scored between 50 to 74")
print(f"{count_below_50} students scored below 50")
```
