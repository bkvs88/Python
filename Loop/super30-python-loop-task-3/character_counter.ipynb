### Count the frequency of every character in a string without using Counter.

```python
string="banana" #initializing a string variable with the value "banana"
dict_op={} #initializing an empty dictionary to store the count of each character in the string
for i in string: # looping through each character in the string
    dict_op[i]=dict_op.get(i,0)+1 #using the get() method to retrieve the current count of the character from the dictionary, if it exists, or 0 if it doesn't exist, and then incrementing the count by 1
for key, value in dict_op.items(): # looping through each key-value pair in the dictionary using the items() method
    print(f"{key} -> {value}") # printing the character and its count in the format "character -> count"
```
