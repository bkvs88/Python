### Create a simple login system with a maximum of 3 password attempts.


```python

correct_pwd="Password" #initializing a variable to store the correct password
max_attempts=3 #initializing a variable to store the maximum number of attempts allowed
for attempt in range(1,max_attempts+1): # looping through the range of attempts from 1 to max_attempts
    user_input=input(f"Enter your passowrd Attempt {attempt}/{max_attempts}")
    if user_input == correct_pwd: # checking if the user input matches the correct password
        print("Welcome to the world of python") 
        break #breaking the loop if the password is correct
    else:
        print("Incorrect password, Try again")
else:
    print("Account Locked, Maximum Attempts Exceeded")
```
