### Create a basic ATM simulation where a user can repeatedly

```python
balance=10000.00
is_running=True

print("Welcome to Worlds Best Bank")

while is_running:
    print("\n =========\n ATM Menu \n =========")
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw money")
    print("4. Exit")

    choice=int(input("Enter your choice between 1 to 4"))

    if choice==1:
        print(f"Your current Bank Balance is {balance}")
    elif choice==2:
        amount=float(input("Enter amount you would like to deposit"))
        if amount>0:
            balance=balance+amount
            print(f"succesfully Deposited {amount}")
            print(f"Latest balance is {balance}")
        else:
            print("Invalid amount , Please provide latest amount to Deposit")
    elif choice==3:
        amount=float(input("Enter amount you would like to withdraw"))
        if amount>0 and amount<=balance:
            balance=balance-amount
            print(f"succesfully withdraw {amount}")
            print(f"Latest balance is {balance}")
        elif amount>balance:
            print(f"Insufficient balance, Your current balance is {balance}")
        else:
            print("Invalid amount , Please provide latest amount to Deposit")
    elif choice==4:
        print("Thank you for visiting ATM")
        is_running=False
    else:
        print("Invalid choice, Select any choice between 1 to 4")
```
