```python
# Input list with exactly 10 elements
numbers = [12, 45, 2, 45, 34, 23, 7, 8, 34, 1]

# Step 1: Remove duplicates to handle repeating highest numbers
unique_numbers = list(set(numbers))

# Step 2: Ensure there are at least two unique elements
if len(unique_numbers) < 2:
    print("The list does not have enough unique numbers.")
else:
    # Initialize trackers to negative infinity
    largest = float('-inf')
    second_largest = float('-inf')

    # Step 3: Loop through elements to find the top two values
    for num in unique_numbers:
        if num > largest:
            second_largest = largest
            largest = num
        elif num > second_largest:
            second_largest = num

    # Output the result
    print("Original list:", numbers)
    print("The second-largest number is:", second_largest)
```
