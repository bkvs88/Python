### Check whether a string is a palindrome using loops.

```python
input_str=input("Enter string to check whether palindrome or not")
is_palindrome=True
n = len(input_str)
#print(n//2)
for i in range(n // 2):
    #print(f"{input_str[i]} , {input_str[n-1-i]}")
    if input_str[i] != input_str[n - 1 - i]:
        is_palindrome = False
        break
if is_palindrome:
    print(f'"{input_str}" is a palindrome string')
else:
    print(f'"{input_str}" is not a palindrome string')
```
