### Reverse a string using a for loop without using [::-1] or reversed().

```python
input_string = "HELLO" #taking input string from the user
reversed_str="" #initializing an empty string to store the reversed string
for i in input_string: #looping through each character in the input string
   reversed_str= i + reversed_str #concatenating the current character to the reversed string
print(reversed_str)
```
