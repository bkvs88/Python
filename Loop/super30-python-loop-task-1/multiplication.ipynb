### Take an integer n and print its multiplication table from 1 to 20.

```python
num_inp = int(input("enter any integer number that need table from 1 to 20")) #taking input from the user
print(f"Displaying {num_inp} table ") #printing the table name
for i in range(1,21): #looping through the range 1 to 20
    val=num_inp*i #calculating the product of the input number and the current iteration
    print(f"{num_inp} X {i} = {val}") 
```
