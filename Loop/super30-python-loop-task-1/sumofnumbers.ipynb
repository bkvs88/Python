### Calculate the sum of numbers from 1 to n using a loop.

```python
num_inp = int(input("enter any number that need sum from 1 to n")) #taking input from the user
i =1 #initializing i variable to 1
sum=0 #initializing sum variable to 0
while i <=num_inp :
    sum = sum+i
    i=i+1

print(f"Total sum from 1 to {num_inp} is {sum}")
```
