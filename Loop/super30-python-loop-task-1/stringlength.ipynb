### Given languages = ["Python", "Java", "C++", "JavaScript", "Go"] Print every language along with its length.

```python
languages = ["Python", "Java", "C++", "JavaScript", "Go"] #initializing a list of programming languages
for i in languages: #looping through each language in the list
    print(f"length of {i} is {len(i)}") #printing the length of each language using len() function
```
