### Iterate through student = {"name": "Rahul","age": 22,"course": "Data Science","city": "Bangalore" } and print every key and value.

### This is a simple python program to print the key-value pairs of a dictionary using for loop and items() method.

```python
#initializing a dictionary with student details
student = {
"name": "Rahul",
"age": 22,
"course": "Data Science",
"city": "Bangalore"
} 
#iterating through the dictionary and printing key-value pairs
for key,value in student.items():
    print(f"{key} : {value} " )
```
