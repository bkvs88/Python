### Count how many vowels exist in a user-provided string.


```python
inp_string=input("Enter string to check number of vowels")
print (f"string entered is {inp_string}")
inp_string=inp_string.lower() # converting input string all to lower case.
vowels = "aeiou" #vowels variable assignment to check 
vowel_count=0 #initiliase the variable to 0
for char in inp_string:
    if char in vowels:
        vowel_count=vowel_count+1
print(f'number of vowels found in ""{inp_string}"" are {vowel_count}')
```
