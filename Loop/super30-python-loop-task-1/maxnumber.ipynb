### Find the largest number from a list without using max().

```python
numbers = [12, 7, 9, 20, 33, 42, 8, 15, 25, 99, 86,-1] #initializing list of numbers
large_number= numbers[0] #initializing large_number variable to the first number in the list to store the largest number from the list
for i in numbers: #looping through the numbers in the list
    if i>large_number: #checking if the current number is greater than the large_number variable
        large_number=i #assigning the current number to the large_number variable if it is greater than the large_number variable
    else:
        large_number=large_number #assigning the large_number variable to itself if the current number is not greater than the large_number variable
print(f'{large_number} is the Largest number from the given list {numbers} ')
```
