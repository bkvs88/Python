### Print all odd numbers from 1 to 100
### using for loop and range()

```python
for i in range(1,101,2): #looping through numbers from 1 to 100 with a step of 2 using range function
    print(i)
```

### using while loop

```python
i=1 #initializing i variable to 1
while i<=100: #looping through numbers from 1 to 100 using while loop
    if i%2==1: #checking if the current number is odd using modulus operator
        print(i)
    i=i+1
```
