### Print all even numbers from 1 to 100.
### using for loop and range()
#### range function generates a sequence of numbers starting from 2 to 100 with a step of 2, which means it will generate even numbers only.

```python
for i in range(2,101,2): #looping through numbers from 2 to 100 with a step of 2 using range function
    print(i)
```

### using while loop

```python
i=1
while i<=100:
    if i%2==0:
        print(i)
    i=i+1
```
