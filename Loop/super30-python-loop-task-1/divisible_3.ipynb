```python
#initializing list of numbers
numbers = [12, 7, 9, 20, 33, 42, 8, 15] 
print(f"Following numbers are divisible by 3 for given list {numbers}")
div_by_3_list=[] #initializing empty list to store numbers divisible by 3
for i in numbers:
    if i%3==0:
        div_by_3_list.append(i) #appending numbers divisible by 3 to the list
print(f"Numbers divisible by 3: {div_by_3_list}")
```
