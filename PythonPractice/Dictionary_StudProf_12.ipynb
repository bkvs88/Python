## print all keys using keys()
### keys() method returns a view object that contains all the keys present in a dictionary.

```python
student.keys()
```

## print all values using values()
### values() method returns a view object containing all the values stored in a dictionary.

```python
student.values()
```

## print all key value pairs using items()
### items() method returns a view object containing the key-value pairs of a dictionary as tuples.

```python
student.items()
```

## Use get() to retrieve "name"

## Create a copy of the dictionary using copy()
### copy() method creates a shallow copy of a collection, This ensures that modifications made to the copy do not alter the original collection.

```python
new_stud=student.copy()
```

```python
new_stud
```

```python
student["city"]="Hyderabad"
```

```python
student, new_stud
```
