## Convert the list into a 
### set() in python is an unordered collection of unique items. 
### set() to remove duplicates. Sets are very useful when working with duplicate data because they handle deduplication automatically and perform data checks with extreme speed.

```python
set_numbers = set(numbers)
```

```python
type(set_numbers)
```

```python
len(set_numbers)
```

## Observe which duplicates disappear
### duplicates 10, 20, 30 elemets are disappeared.

```python
set_numbers
```

```python
len(set_numbers)
```

## Convert the set back into a list

```python
uniq_list_numbers=list(set_numbers)
```

```python
print(uniq_list_numbers)
```
