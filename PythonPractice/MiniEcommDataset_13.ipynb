### List: A mutable, ordered sequence of elements enclosed in square brackets 
### Dictionary: A mutable collection of unique key-value pairs enclosed in curly braces
### Indexing: Accessing a single element in a sequence using its position number inside square brackets, like my_list[0].
### Keys: The unique identifiers in a dictionary used to look up data, extracted using the .keys() method.
### Values: The actual data associated with keys inside a dictionary, extracted using the .values() method.
