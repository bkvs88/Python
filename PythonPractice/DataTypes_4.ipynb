## Rules to Follow for variable naming convention
#### Should start with Underscore or characters (Not start with numbers or any special character other than underscore)
#### No reserved keywords shoule use as variable llike (if , else, while, class, import etc..)
##### To use lowercase letters for variables (best practice)
