## Defining a tuple variable
### Tuples are called immutable because their state cannot be modified after they are created. Once a tuple is defined in memory, you cannot add, remove, or change its elements in place.

```python
technologies = ("Python", "Java", "Python", "C++", "JavaScript", "Python")
```

## Print the tuple

```python
technologies
```

## Print type of the variable

```python
type(technologies)
```

## Print the first item using index

```python
technologies[0]
```

## Print the last item using index

```python
technologies[-1]
```

## Slice the Tuple using index start and end position

```python
technologies[0:]
```

```python
technologies[0:2]
```

```python
technologies[::-1]
```

## Count Python using count()

```python
technologies.count("Python")
```

## Find the Index of C++ using index()

```python
technologies
```

```python
technologies.index("C++")
```

## Find the tuple Length using len

```python
len(technologies)
```

## Convert the tuple into a list using list()

```python
ls_tech = list(technologies)
```

```python
ls_tech
```

## Add "Go" after converting it into a list using append

```python
ls_tech
```

```python
ls_tech.append("Go")
```

```python
ls_tech
```

## Convert it back to tuple using tuple()

```python
tp_tech = tuple(ls_tech)
```

```python
type(tp_tech)
```
