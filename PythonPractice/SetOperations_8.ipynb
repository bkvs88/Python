## Defining Variables

```python
python_students = {"Rahul", "Aman", "Priya", "Karan", "Neha"}

java_students = {"Priya", "Karan", "Rohit", "Simran"}
```

## Find students learning either Python or Java using union()
### union() method combines two or more sets and returns a new set containing all unique elements from the involved collections

```python
python_students.union(java_students)
```

## Find students learning both using intersection()
### intersection() method returns a new set containing only the elements that exist in all involved collections.

```python
python_students.intersection(java_students)
```

## Find students learning only Python using difference()
### difference() method compares sets and returns a new set containing elements that exist in the first set but NOT in the others

```python
python_students.difference(java_students)
```

## Find students learning only Java

```python
java_students.difference(python_students)
```

## Find students belonging to exactly one group using symmetric_difference()
### symmetric_difference() method compares two collections and returns a new set containing elements that are in either of the sets, but NOT in both

```python
python_students.symmetric_difference(java_students)
```

## Remove a Student using remove()
### remove() method deletes a specific element from an existing set.

```python
python_students.remove("Priya")
```

## Remove a student using discard()
### discard() method removes a specific element from an existing set. Unlike remove(), if the element does not exist in the set, discard() does nothing and does not raise an error.

```python
java_students.discard("Neha")
```
