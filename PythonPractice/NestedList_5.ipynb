## Print Rahul's name using Indexing with nested Lists
#### Python indexing nested lists works by chaining square brackets [] together. first index output will be input to second and so on. 
##### from the given example, first index students[0] will extract Rahuls record list, studetns[0][0] will extract Rahul from list

```python
students[0][0]
```
