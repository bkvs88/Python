## Create the String with Variable

```python
message = " Welcome To Python Programming Class "
```

## Remove Extra Space using Strip() function
#### strip() method in Python returns a copy of a string with both leading and trailing characters removed, By default it removes while spaces from the very beggining and the very end of the text

```python
message=message.strip()
```

## Convert message string to Upper Case using Upper()
#### upper() method in pythond returns all characters converted to uppercase

```python
message.upper()
```

## Convert message string to Lower Case using Lower()
#### lower() method in pythond returns all characters converted to lowercase

```python
message.lower()
```

## Convert message string to title case using title()
#### title() method in python returns first character of every word is Capitalized

```python
message.title()
```

```python
message.isdigit()
```

## Replace Python with "Advanced Python" in message string
#### replace() method in Python returns a copy of a string where all occurrences of a specific substring are substituted with a new substring.

```python
message.replace('Python','Advanced Python')
```

## Check Whether message string starts with Welcom using startswith() 
#### startswith() method in Python returns True if a string begins with a specified prefix; otherwise, it returns False.

```python
message.startswith("Welcome")
```

## Check Whether message string ends with Class using endswith()
#### endswith() method in Python returns True if a string ends with a specified suffix; otherwise, it returns False.

```python
message.endswith("Class")
```

## Count occurences of o in message string using count() 
#### count() method in Python returns the total number of times a specific substring appears in a string.

```python
message.count("o")
```

## Find the position of "Programming" in message input string using find() and index()
#### find() method in Python searches a string for a specific substring and returns the lowest index (the position) where it is first found.

```python
message.find("Programming")
```

```python
message.index("Programming")
```

## Split the sentence into Words using split()
#### split() method in Python breaks a single string into a list of smaller strings based on a specified delimiter (separator).

```python
message.split()
```
