## Insert "USB Hub" at index2 using insert() .
### difference between append(),insert() and extend() .
#### insert() adds an items at a specific position .
#### append() always insert the element to last position in list . adds to the end as a single item.
#### extend() merges items into the end, extend() method takes a collection and appends each individual element one by one to the end of your original list . 

```python
cart.insert(2,"USB Hub")
```

```python
cart
```

## Remove Mouse using remove()
### difference between remove and pop()
#### remove() deletes a specific element by its value (or key/set item) and does not return anything.
#### pop() deletes an element by its index (or key/randomly) and explicitly returns the removed item.

```python
cart.remove("Mouse")
```

```python
cart
```

```python
cart.pop()
```

```python
cart
```
