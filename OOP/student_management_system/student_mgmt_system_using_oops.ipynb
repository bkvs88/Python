Student Management System

```python
class Student:
    # Class Variable to track the total number of students
    total_students = 0

    def __init__(self, name, email, student_id, course, marks=None):
        # Instance Variables
        self.name = name
        self.email = email
        self.student_id = student_id
        self.course = course
        # Default to an empty list if no marks are provided
        self.marks = marks if marks is not None else []
        
        # Increment the class variable every time a new student is created
        Student.total_students += 1

    # Instance Method: Display student details
    def display_details(self):
        print(f"\n--- Student Report: {self.name} ---")
        print(f"ID:      {self.student_id}")
        print(f"Email:   {self.email}")
        print(f"Course:  {self.course}")
        print(f"Marks:   {self.marks}")
        print(f"Average: {self.calculate_average():.2f}")

    # Instance Method: Update / Add new marks
    def update_marks(self, new_marks):
        if isinstance(new_marks, list):
            self.marks.extend(new_marks)
        else:
            self.marks.append(new_marks)
        print(f"--> Marks updated for {self.name}.")

    # Instance Method: Calculate average marks
    def calculate_average(self):
        if not self.marks:
            return 0.0
        return sum(self.marks) / len(self.marks)

    # Class Method: Accesses class-level data (uses 'cls' instead of 'self')
    @classmethod
    def get_total_students(cls):
        return f"Total students registered: {cls.total_students}"
```

```python
# 1. Create multiple student objects
student1 = Student("Sravan", "sravan@mail.com", "STU001", "Computer Science", [85, 90, 78])
student2 = Student("Kumar", "kumar@mail.com", "STU002", "Data Science", [92, 88, 95])
student3 = Student("Kesav", "kesav@mail.com", "STU003", "Cyber Security") # No marks initially

# 2. Display initial details
student1.display_details()
student3.display_details() # Average will be 0.0

# 3. Update marks for a student
student3.update_marks([75, 80, 83])
student3.display_details() # Showing new average

# 4. Use the class method to see the total headcount
print("\n------------------------------------")
print(Student.get_total_students())
```

## Understanding the Student Class — OOP Concepts Explained

### 1. Class
A **class** is a blueprint. `class Student` defines the structure for all student objects — it says what attributes a student has and what actions (methods) a student can perform.

### 2. Objects
We created three **objects** — `student1`, `student2`, and `student3`. Each is a unique student built from the same `Student` blueprint, but each holds its own data.

### 3. `__init__` Method
The `__init__` method is the **constructor**. It runs the moment you create an object:
- It assigns `name`, `email`, `student_id`, `course`, and `marks` to the specific object using `self`.
- It increments `Student.total_students` so the class always knows how many students exist.
- The `marks=None` default lets you create a student even without marks.

### 4. Instance Methods
These are methods that belong to each **object** (identified by `self`):
- **`display_details()`** — Prints the student's info and calculated average.
- **`update_marks(new_marks)`** — Adds new marks to the student's list.
- **`calculate_average()`** — Returns the average of the student's marks.

Each object calls these methods on its own data. `student1.display_details()` prints Sravan's data, not Kesav's.

### 5. Class Variable
`total_students = 0` is a **class variable** — it belongs to the `Student` class itself, not to any single object. Every time a new object is created in `__init__`, this counter increases. All objects and the class share this single value.

### 6. Class Method
The method `get_total_students()` is decorated with `@classmethod` and uses `cls` instead of `self`:
- It accesses `cls.total_students` (the class variable).
- It can be called directly on the class: `Student.get_total_students()` — no object needed.
- It gives a summary of class-level data (total registered students).

---

### Key Takeaway
| Concept | What it is | Example |
|---|---|---|
| Class | Blueprint for objects | `class Student` |
| Object | Instance of a class | `student1 = Student(...)` |
| `__init__` | Constructor, initializes attributes | `def __init__(self, name, ...)` |
| Instance Method | Method operating on an object | `display_details()`, `calculate_average()` |
| Class Variable | Shared across all instances | `total_students = 0` |
| Class Method | Operates on the class itself | `get_total_students()` |
