Student Management System

```python
class Student:
    # Class Variable to track the total number of students
    total_students = 0

    def __init__(self, name, email, student_id, course, marks=None):
        # Instance Variables
        self.name = name
        self.email = email
        self.student_id = student_id
        self.course = course
        # Default to an empty list if no marks are provided
        self.marks = marks if marks is not None else []
        
        # Increment the class variable every time a new student is created
        Student.total_students += 1

    # Instance Method: Display student details
    def display_details(self):
        print(f"\n--- Student Report: {self.name} ---")
        print(f"ID:      {self.student_id}")
        print(f"Email:   {self.email}")
        print(f"Course:  {self.course}")
        print(f"Marks:   {self.marks}")
        print(f"Average: {self.calculate_average():.2f}")

    # Instance Method: Update / Add new marks
    def update_marks(self, new_marks):
        if isinstance(new_marks, list):
            self.marks.extend(new_marks)
        else:
            self.marks.append(new_marks)
        print(f"--> Marks updated for {self.name}.")

    # Instance Method: Calculate average marks
    def calculate_average(self):
        if not self.marks:
            return 0.0
        return sum(self.marks) / len(self.marks)

    # Class Method: Accesses class-level data (uses 'cls' instead of 'self')
    @classmethod
    def get_total_students(cls):
        return f"Total students registered: {cls.total_students}"
```

```python
# 1. Create multiple student objects
student1 = Student("Sravan", "sravan@mail.com", "STU001", "Computer Science", [85, 90, 78])
student2 = Student("Kumar", "kumar@mail.com", "STU002", "Data Science", [92, 88, 95])
student3 = Student("Kesav", "kesav@mail.com", "STU003", "Cyber Security") # No marks initially

# 2. Display initial details
student1.display_details()
student3.display_details() # Average will be 0.0

# 3. Update marks for a student
student3.update_marks([75, 80, 83])
student3.display_details() # Showing new average

# 4. Use the class method to see the total headcount
print("\n------------------------------------")
print(Student.get_total_students())
```
